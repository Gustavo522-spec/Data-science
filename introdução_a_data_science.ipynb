```python
total_por_lingua = tmdb["original_language"].value_counts()
total_geral = total_por_lingua.sum()
total_de_ingles = total_por_lingua.loc["en"]
total_do_resto = total_geral - total_de_ingles
print(total_geral, total_de_ingles, total_do_resto)
```

```python
dados = {
"lingua" : ["ingles", "outros"],
"total" : [total_de_ingles, total_do_resto]
}
dados = pd.DataFrame(dados)
dados
```

```python
sns.barplot(data=dados, x="lingua", y="total")
```

```python
dados.plot(kind="pie", y="total", labels=dados["lingua"])
```

```python
total_de_outros_filmes_por_lingua = tmdb.query("original_language != 'en'")["original_language"].value_counts()
total_de_outros_filmes_por_lingua.head()
```

```python
sns.countplot(data=tmdb.query("original_language != 'en'"),
            x="original_language")
```

```python
sns.countplot(data=tmdb.query("original_language != 'en'"),
              order=total_de_outros_filmes_por_lingua.index,
            x="original_language")
```

```python
plt.figure(figsize=(16,8))
sns.countplot(data=tmdb.query("original_language != 'en'"),
              order=total_de_outros_filmes_por_lingua.index,
              hue="original_language",
            x="original_language")

#ordem de cores meio desorganizada, o que dificulta na compreenção
```

```python
sns.color_palette("mako")

#essa paleta de cores facilitara na visualização e interpretação desse grafico
```

```python
plt.figure(figsize=(16,8))
sns.countplot(data=tmdb.query("original_language != 'en'"),
              order=total_de_outros_filmes_por_lingua.index,
              hue="original_language",
              palette="mako",
              hue_order=total_de_outros_filmes_por_lingua.index,
              stat="percent",
            x="original_language")
plt.title("Distribuição da lingua original nos filmes exceto em inglês")
plt.show()

#aqui usaremos a paleta de cores que esta acima, junto tambem colocaremos a porcemtagem que cada lingua representa.
```

```python
import numpy as np

filme1 = [2.5] * 10 + [3.5] * 10
np.mean(filme1)
```

```python
filme2 = [5] * 10 + [1] * 10
np.mean(filme2)
```

```python
media_filme1 = np.mean(filme1)
mediana_filme1 = np.median(filme1)

media_filme2 = np.mean(filme2)
mediana_filme2 = np.median(filme2)

print("filme 1:")
print("media:", media_filme1)
print("mediana:", mediana_filme1)

print("filme 2:")
print("media:", media_filme2)
print("mediana:", mediana_filme2)

```
